```python
# Crawl Data

filename = 'palestinaa.csv'
search_keyword = 'palestina lang:id until:2023-10-31 since:2023-01-01'
limit = 100

!npx --yes tweet-harvest@latest -o "{filename}" -s "{search_keyword}" -l {limit} --token ""
```

```python
import pandas as pd

# Specify the path to your CSV file
file_path = f"tweets-data/{filename}"

# Read the CSV file into a pandas DataFrame
df = pd.read_csv(file_path, delimiter=";")

# Select only the desired columns 'created_at', 'full_text', and 'username'
df = df[['created_at', 'full_text', 'username']]

# Display the DataFrame
display(df)
```
